## Lec 9  - Cleaning Data

```python
import pandas as pd
import numpy as np
```

```python
people = {
    'first': ['Corey', 'Jane', 'John', 'Chris', np.nan, None, 'NA'], 
    'last': ['Schafer', 'Doe', 'Doe', 'Schafer', np.nan, np.nan, 'Missing'], 
    'email': ['CoreyMSchafer@gmail.com', 'JaneDoe@email.com', 'JohnDoe@email.com', None, np.nan, 'Anonymous@email.com', 'NA'],
    'age': ['33', '55', '63', '36', None, None, 'Missing']
}
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
# drop values with na (np.na)  - will not drop string "Na"

# df.dropna() 

# df.dropna(axis='index',how='any') # same as above, any - if any column is na
#axis = drop index or col (if columns),  how - criteria to drop

# df.dropna(axis='columns',how='all')  

df.dropna(axis='index',how='all') # all - drop only if all values are missing in row
```

```python
# drop rows without email

# df.dropna(axis='index', how='any', subset=['email']) 
# df.dropna(axis='index', how='all', subset=['last','email']) # drop if both email and last missing

df.dropna(axis='index', how='any', subset=['last','email']) # drop if  either of  email and last missing
```

```python
# to chaneg string NA to numpy naabs
df.replace('NA',np.nan)
```

```python

# to chaneg string NA to numpy naabs
df.replace('NA',np.nan, inplace=True)
df.replace('Missing',np.nan, inplace=True)
```

```python
df
```

```python
df.dropna()
```

```python
df.dropna(axis='index', how='any', subset=['last','email'])
```

```python
# check values which are na
df.isna()
```

```python
# to fill na value as MISSING

# df.fillna('MISSING')
df.fillna(0)
```
