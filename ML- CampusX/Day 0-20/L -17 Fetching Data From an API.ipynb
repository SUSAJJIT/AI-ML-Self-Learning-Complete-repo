```python
# 1. Create an empty list to store the dataframes from each page
df_list = []

for i in range(1, 429):
    url = 'https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={}'.format(i)
    response = requests.get(url)

    # 2. Extract the data and create a temporary DataFrame
    data = response.json().get('results', [])
    if data:
        temp_df = pd.DataFrame(data)[['id', 'title', 'overview', 'release_date', 'popularity']]
        # 3. Add this specific page's data to our list
        df_list.append(temp_df)

# 4. Concatenate everything at once (much more efficient)
df = pd.concat(df_list, ignore_index=True)
```
